# arancel-mx — Audit & NICO Coverage Example

This notebook demonstrates verified access to the public Mexican tariff dataset.

**Always cite the exact `data-YYYY.MM.DD` release.**

````python
from arancel_mx import Dataset

ds = Dataset.latest()
print(ds.info.dataset_tag, "verified:", ds.info.release_verified)
````

````python
# Exact lookup + hierarchy
code = "85171301"
rec = ds.lookup(code)
print(f"{rec.code} — {rec.description}")
print(f"IGI: {rec.igi_text} | IGE: {rec.ige_text}")

print("\nHierarchy:")
p = ds.parent(rec.code)
while p:
    print(f"  {p.level}: {p.code}")
    p = ds.parent(p.code)
````

````python
# Bulk search example
hits = ds.search("refrigerador", limit=5)
for h in hits:
    print(h.record.code, h.record.description[:60])
````
