# Auditoría de una release verificada

Este cuaderno usa solo `Dataset`. No clasifica mercancías y no inventa NICO ni fechas.

Cita `dataset_version`, si la release está verificada, y el archivo local.

```python
from arancel_mx import Dataset

dataset = Dataset.latest()
info = dataset.info
print(info.dataset_version, info.schema_version, info.release_verified, info.source)
print(info.path)
```

```python
record = dataset.lookup("85171301")
print(record.code, record.level, record.description)
print("IGI", record.igi_text, "IGE", record.ige_text)
print("basis", record.validity_basis, "from", record.effective_from, "to", record.effective_to)

card = dataset.ficha(record.code)
print("hierarchy", [item.code for item in card.hierarchy])
print("children", len(card.children))
```

```python
for hit in dataset.search("telefono", limit=5):
    print(hit.match_kind, hit.confidence, hit.record.code, hit.record.description[:70])
```
